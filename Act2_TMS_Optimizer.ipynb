```python
#Representación del conocimiento: Grafo de transportes con base en reglas de conexión y distancias
freight_network = {
    'Portal Norte': {'Calle 100': 2, 'Suba': 5},
    'Calle 100': {'Portal Norte': 2, 'Suba': 1, 'Portal 80': 8},
    'Suba': {'Portal Norte': 5, 'Calle 100': 1, 'Portal 80': 3},
    'Portal 80': {'Calle 100': 8, 'Suba': 3}
}
```

```python
# 2. Heurística

heuristica = {
    'Portal Norte': 0,
    'Calle 100': 2,
    'Suba': 3,
    'Portal 80': 4
}
```

```python
import heapq

# 2. Heurística
# Definition copied here to resolve NameError if the preceding cell was not executed.
heuristica = {
    'Portal Norte': 0,
    'Calle 100': 2,
    'Suba': 3,
    'Portal 80': 4
}

#Se define la función  estandar a_star para priorizar la mejor ruta de transporte
def a_star(graph, start, goal, heuristica):
  queue_priority=[(0, start)]
  came_from={start: None}
  cost_so_far={start: 0}

  while queue_priority:
    current_priority, current_node = heapq.heappop(queue_priority)

    if current_node == goal:
      break
    # Iterate over items to get next_node and cost
    for next_node, cost in graph[current_node].items():
      new_cost = cost_so_far[current_node] + cost
      if next_node not in cost_so_far or new_cost < cost_so_far[next_node]:
        cost_so_far[next_node] = new_cost
        priority = new_cost + heuristica[next_node]
        heapq.heappush(queue_priority, (priority, next_node))
        came_from[next_node] = current_node

  # Recalcular ruta - this part should be executed AFTER the main loop
  route=[]
  if goal not in came_from:
      return [] # Return empty list if goal is not reachable
  curr = goal
  while curr != start:
    route.append(curr)
    curr = came_from[curr]
  route.append(start)
  route.reverse()
  return route # Corrected syntax

route_x = a_star(freight_network, 'Portal Norte', 'Portal 80', heuristica)
print(f"La mejor ruta encontrada desde Portal Norte hasta Portal 80 es: {route_x}")
```
